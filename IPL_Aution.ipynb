```python
sns.histplot(df["player style"])
plt.title("Player Style Distribution")
plt.xlabel("Player Style")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.histplot(df["franchise"])
plt.title("Franchise Distribution")
plt.xlabel("Franchise")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.tight_layout()      
plt.show()  
```

```python
columns_to_plot = ["nationality", "status"]

num_rows = 1
num_cols = 2

fig, axes = plt.subplots(nrows=num_rows, ncols=num_cols, figsize=(12, 6))

axes = axes.flatten()

for i, column in enumerate(columns_to_plot):
    sns.histplot(df[column], ax=axes[i])
    axes[i].set_title(f"Distribution of {column.capitalize()}")
    axes[i].set_xlabel(column.capitalize())
    axes[i].set_ylabel("Count")
    axes[i].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df["base price (in lacs)"], df["final price (in lacs)"], color='blue', alpha=0.5)
plt.title("Base Price vs Sold Price")
plt.xlabel("Base Price")
plt.ylabel("Sold Price")
plt.show()  
```
